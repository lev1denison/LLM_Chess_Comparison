# Local-only study setup

The study now uses `qwen3:8b`, `llama3.1:8b`, and `gemma3:4b` through Ollama. No API key is required. Install with `ollama pull MODEL_NAME` and keep Ollama running.

Run this notebook with the project .venv kernel. The check below lists installed models without inference. Collection and analysis are separate: `run_llm_puzzles.py` and `early_results.ipynb`.

The shared prompt and frozen settings are in `chess_prompt_v1.txt` and `llm_config.json`. Qwen thinking is disabled for the direct-answer experiment; local model digests and quantization are recorded in the experiment manifest.
